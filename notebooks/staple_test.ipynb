```python
# get grid potential via potential_and_force
N = 200
xvals = np.linspace(0., 0.92, N)
yvals = np.linspace(-0.01, 0.31, N)
potvals = np.zeros((N, N))
for i, x in enumerate(xvals):
    for j, y in enumerate(yvals):
        potvals[j, i], _ = sim.ensembles[0].engine.potential.potential_and_force(
            (np.array([y, x]), np.array([0., 0.]))
        )

fig, ax = plt.subplots(figsize=(8, 6))

vmin = np.min(potvals)
vmax = np.max(potvals)

cs = ax.contourf(xvals, yvals, potvals, 100, cmap="viridis", vmin=vmin, vmax=vmax, extend="both")

# add interfaces
for k, intf in enumerate(sim.intfs):
    ax.axvline(intf, color="white", linestyle="--", linewidth=1.2, alpha=0.9,
               label="interfaces" if k == 0 else None)

ax.set_title("Potential surface (potential_and_force)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend(loc="upper right")

fig.colorbar(cs, ax=ax, label="potential", extend="both")
plt.show()
```
