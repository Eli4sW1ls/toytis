```python
# plot the potential and the interfaces
x = np.linspace(-0.3, 1.5, 1000)
pot = np.zeros_like(x)
for i in range(len(x)):
    pot[i], _ = sim.ensembles[0].engine.potential.potential_and_force((x[i], 0.))

fig,ax =plt.subplots()
ax.plot(x, pot)
for intf in sim.intfs:
    ax.axvline(intf, color="black", linestyle="--")
fig.show()
```
